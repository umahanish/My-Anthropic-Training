```python
# create an API client
import os
from pathlib import Path

from anthropic import Anthropic
from dotenv import load_dotenv

workspace_root = Path(r"C:/AI-Training/Anthropic-learning/My-Anthropic-Training")
dotenv_path = workspace_root / ".env"
load_dotenv(dotenv_path, override=True)

api_key = os.getenv("ANTHROPIC_API_KEY")
if not api_key:
    raise ValueError("ANTHROPIC_API_KEY is not set. Add it to your .env file.")
if len(api_key) < 20:
    raise ValueError("ANTHROPIC_API_KEY looks too short to be valid.")

client = Anthropic(api_key=api_key)
model = "claude-sonnet-4-5"
print("Client ready")
```

```python
def add_user_message(messages, text):
    user_message = {"role": "user", "content": text}
    messages.append(user_message)

def add_assistant_message(messages, text):
    assistant_message = {"role": "assistant", "content": text}
    messages.append(assistant_message)

def chat(messages, system=None):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
    }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
messages = []

system = """""
you are a patient math tutor.
Do not directly answer a student's question.
Guide them to a solution step by step.
"""""
add_user_message(messages, "How do I Solve 5x+3=2 for x?")
answer = chat(messages, system=system)

answer
```

```python
def chat(messages, system=None, temperature=1.0):
    params = {
        "model": model,
        "max_tokens": 1000,
        "messages": messages,
        "temperature": temperature
    }
    
    if system:
        params["system"] = system
    
    message = client.messages.create(**params)
    return message.content[0].text
```

```python
prompt = "Invent a short, unusual name and one-sentence description for a fictional underwater library."
temperatures = [step / 10 for step in range(11)]

for temperature in temperatures:
    messages = [{"role": "user", "content": prompt}]
    answer = chat(messages, temperature=temperature)
    print(f"\nTemperature {temperature:.1f}:\n{answer}")
```
